```python
#By: Lara Jamile Galvão, Larah Sofia Rabelo e Heitor Guinzani de Vasconcelos

gastos = []

orcamento = float(input("Qual o seu orçamento? R$ "))
rendaMensal = float(input("Qual sua renda mensal? R$ "))

while True:

    print("\nCategorias:")
    print("1 - Alimentação")
    print("2 - Lazer")
    print("3 - Transporte")
    print("4 - Saúde")
    print("5 - Educação")
    print("6 - Outros")

    categoria = input("Escolha a categoria: (Apenas números) ")

    nomeProduto = input("\nQual o nome do produto? ")

    valorProduto = float(input("Qual o valor do gasto? R$ "))

    if categoria == "1":
        categoria = "Alimentação"
    elif categoria == "2":
        categoria = "Lazer"
    elif categoria == "3":
        categoria = "Transporte"
    elif categoria == "4":
        categoria = "Saúde"
    elif categoria == "5":
        categoria = "Educação"
    elif categoria == "6":
        categoria = "Outros"
    else:
        categoria = "Outros"

    gastos.append([nomeProduto, categoria, valorProduto])

    outro = input("Há outros gastos? (sim/não): ").lower()

    if outro == "não":
        break

total = 0

print("\n--- RELATÓRIO DE GASTOS ---")

for gasto in gastos:
    print(f"Produto: {gasto[0]}")
    print(f"Categoria: {gasto[1]}")
    print(f"Valor: R$ {gasto[2]:.2f}")
    print("----------------------")

    total += gasto[2]

print(f"\nRenda mensal: R$ {rendaMensal:.2f}")
print(f"Orçamento: R$ {orcamento:.2f}")
print(f"Total gasto: R$ {total:.2f}")

saldo = orcamento - total

print(f"Saldo do orçamento: R$ {saldo:.2f}")

if total > orcamento:
    print("Você ultrapassou o orçamento!")
else:
    print("Você está dentro do orçamento!")
```
